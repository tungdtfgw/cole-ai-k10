```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
f = lambda x: x**2 - 2*x + 3
```

```python
x = np.linspace(-3, 5, 200)

plt.plot(x, f(x))
plt.title('Hàm bậc 2')
plt.show()
```

```python
f = lambda x, y: x**2 + 2*y**2

xs = np.linspace(-3, 3, 100)
ys = np.linspace(-3, 3, 100)
x, y = np.meshgrid(xs, ys)
z = f(x, y)

fig = plt.figure(figsize=(5, 5))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, z, cmap='viridis')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
plt.show()
```

```python
sigmoid = lambda x: 1.0 / (1.0 + np.exp(-x))

x = np.linspace(-15, 15, 100)
y = sigmoid(x)

plt.plot(x, y)
plt.title('Hàm sigmoid')
plt.show()
```

```python
relu = lambda x: np.maximum(0, x)

x = np.linspace(-15, 15, 100)
y = relu(x)

plt.plot(x, y)
plt.title('Hàm ReLU')
plt.show()
#
```

```python
d_sigmoid = lambda x: sigmoid(x) * (1 - sigmoid(x))

x = np.linspace(-15, 15, 100)
y = d_sigmoid(x)

plt.plot(x, y)
plt.title('Đạo hàm hàm sigmoid')
plt.show()
```

```python
d_relu = lambda x: np.where(x > 0, 1, 0)

x = np.linspace(-15, 15, 100)
y = d_relu(x)

plt.plot(x, y)
plt.title('Đạo hàm hàm ReLU')
plt.show()
```

```python
f = lambda x: x**2
df = lambda x: 2*x

x0 = 3
fx0 = f(x0)
dfx0 = df(x0)

print(f'f({x0}) = {fx0}')
print(f'f\'({x0}) = {dfx0}')
```

```python
x = np.linspace(0, 5, 100)
tiep_tuyen = fx0 + dfx0 * (x - x0)

plt.plot(x, f(x), label='f(x)')
plt.plot(x, tiep_tuyen, label='tiếp tuyến')
plt.legend()
plt.show()
```

```python
f = lambda x, y: x**2 + 2*y**2
grad_f2 = lambda x, y: np.array([2*x, 4*y])
d = lambda g: np.linalg.norm(g)

points = [(3, 1), (1, 1), (0.5, 0), (0, 0)]
for px, py in points:
  g = grad_f2(px, py)
  print(f'({px}, {py}): gradient: {d(g):.2f}')
```

```python
p = np.array([3.0, 1.0])
g = grad_f2(*p)
u_grad = g / np.linalg.norm(g)

huong = [u_grad, 0-u_grad, [1.0, 0.0], [0.0, 1.0], np.array([-u_grad[1], u_grad[0]])]

for h in huong:
  print(f'{h}: {g @ h}')
```

```python
J = lambda w: w**2 - 6*w + 10
dJ = lambda w: 2*w - 6

def gradient_descent(w0, alpha, n):
  w = w0
  w_list = [w0]
  j_list = [J(w0)]
  e = 0.001
  for i in range(n):
    g = dJ(w)
    w = w - alpha * g
    print(f't = {i}: w = {w:.2f}, J(w) = {J(w):.2f}')

    if abs(J(w) - j_list[-1]) < e:
      break

    w_list.append(w)
    j_list.append(J(w))

  return i, w_list, j_list
```

```python
n = 1000
alpha = 0.1
w0 = 20
t, ws, Js = gradient_descent(w0, alpha, n)
print(f'Sau {t} lần lặp, w = {ws[-1]:.2f}, J(w) = {Js[-1]:.2f}')
```
