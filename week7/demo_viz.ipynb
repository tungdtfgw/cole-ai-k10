```python
x = np.linspace(0, 10, 101)
y = np.sin(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.title('Sine Wave')
plt.show()
```

```python
def my_plot(x, y, title, xlabel, ylabel):
    plt.plot(x, y)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(title)
    plt.show()
```

```python
y = 2 * x + 5
my_plot(x, y, 'y = 2x + 5', 'x', 'y')
```

```python
products = ['Iphone 17', 'Samsung Galaxy Z', 'One Plus 11', 'Redmi Note 14']
sales = [1000, 1500, 800, 1200]

bars = plt.bar(products, sales, color=['skyblue', 'salmon', 'violet', 'lightgreen'])
plt.bar_label(bars)
plt.title('Products vs Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

```python
colors = plt.cm.rainbow(np.linspace(0, 1, len(products)))
plt.pie(sales, labels=products, autopct='%1.1f%%', colors=colors, startangle=45, explode=(0, 0.1, 0, 0))
plt.title('Products vs Sales')
plt.show()
```

```python
n_scores=500
reading_scores = np.random.normal(loc=70,   # trung bình = 70
                                  scale=10, # std = 10
                                  size=n_scores)
writing_scores = reading_scores + np.random.normal(loc=10, scale=5, size=n_scores)
plt.scatter(reading_scores, writing_scores)
plt.title('Reading vs Writing')
plt.xlabel('Reading Scores')
plt.ylabel('Writing Scores')
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(reading_scores, kde=True, color='cyan', bins=20)
plt.title('Reading score distribution')
plt.xlabel('Reading Scores')
plt.ylabel('Frequency')
plt.show()
```

```python
# Ex: vẽ phân phối writing scores => có phải phân phối chuẩn không?
plt.figure(figsize=(10, 6))
sns.histplot(writing_scores, kde=True, color='purple', bins=20)
plt.title('Reading score distribution')
plt.xlabel('Reading Scores')
plt.ylabel('Frequency')
plt.show()
```

```python
import pandas as pd

df_peinguins = pd.read_csv('penguins_simple.csv')
df_peinguins.head()
```

```python
df_peinguins.info()
```

```python
df_peinguins.isnull().sum()
```

```python
df_peinguins.dropna(inplace=True)
df_peinguins.isnull().sum()
```

```python
df_peinguins.describe()
```

```python
bill_length = df_peinguins['bill_length_mm']
bill_detpth = df_peinguins['bill_depth_mm']
print(bill_length.shape, bill_detpth.shape)
```

```python
plt.scatter(bill_length, bill_detpth)
plt.title('Bill length vs Bill depth')
plt.xlabel('Bill length')
plt.ylabel('Bill depth')
plt.show()
```

```python
flipper_length = df_peinguins['flipper_length_mm']

plt.scatter(bill_length, flipper_length, color='lightblue')
plt.title('Bill length vs Flipper length')
plt.xlabel('Bill length')
plt.ylabel('Flipper length')
plt.show()
#
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_peinguins, x='bill_length_mm', y='flipper_length_mm', hue='species')
plt.title('Bill length vs Flipper length')
plt.xlabel('Bill length')
plt.ylabel('Flipper length')
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_peinguins, x='bill_length_mm', y='body_mass_g', hue='species')
plt.title('Bill length vs Body mass')
plt.xlabel('Bill length')
plt.ylabel('Body mass')
plt.show()
```

```python
mean_body_mass = df_peinguins.groupby('species')['body_mass_g'].mean()
print(mean_body_mass)
```

```python
plt.bar(mean_body_mass.index, mean_body_mass.values)
plt.title('Mean body mass by species')
plt.xlabel('Species')
plt.ylabel('Mean body mass')
plt.show()
```

```python
mean_body_mass = df_peinguins.groupby('island')['body_mass_g'].mean()
print(mean_body_mass)
plt.bar(mean_body_mass.index, mean_body_mass.values)
plt.title('Mean body mass by island')
plt.xlabel('Island')
plt.ylabel('Mean body mass')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df_peinguins, x='species', y='body_mass_g', palette='Set3', hue='species')
plt.title('Body mass by species')
plt.xlabel('Species')
plt.ylabel('Body mass')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
columns = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g', 'species']
sns.pairplot(df_peinguins[columns], hue='species', palette='Set2')
plt.title('Bill length by species')
plt.xlabel('Species')
plt.ylabel('Bill length (mm)')
plt.show()
```

```python
columns = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']
# tạo ma trận tương quan
corr_matrix = df_peinguins[columns].corr()

plt.figure(figsize=(10, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation matrix')
plt.show()
```

```python
corr_matrix_gentoo = df_peinguins[df_peinguins['species'] == 'Gentoo'][columns].corr()

plt.figure(figsize=(10, 6))
sns.heatmap(corr_matrix_gentoo, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation matrix for Gentoo species')
plt.show()
```
