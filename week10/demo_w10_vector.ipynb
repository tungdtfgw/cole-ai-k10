```python
df_raw = np.array([
    [25, 10_000_000],
    [26, 15_000_000],
    [55, 10_500_000]
])
labels = ['John', 'Paul', 'Mike']
```

```python
def d(x, y):
  return np.linalg.norm(x - y)
```

```python
# Tính khoảng cách giữa John và Paul
d_jp = d(df_raw[0], df_raw[1])
d_jm = d(df_raw[0], df_raw[2])
d_pm = d(df_raw[1], df_raw[2])
print(f'John - Paul: {d_jp}')
print(f'John - Mike: {d_jm}')
print(f'Paul - Mike: {d_pm}')
```

```python
Z = (df_raw - df_raw.mean(axis=0)) / df_raw.std(axis=0)
print(Z, Z.mean(axis=0), Z.std(axis=0))
```

```python
print(f'John - Paul: {d(Z[0], Z[1])}')
print(f'John - Mike: {d(Z[0], Z[2])}')
print(f'Paul - Mike: {d(Z[1], Z[2])}')
```
