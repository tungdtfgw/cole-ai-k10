```python
import numpy as np
import pandas as pd
import matplotlib as plt
from scipy import stats
```

```python
# Công ty có mức lương trung bình là 18K (usd), độ lệch chuẩn là 3K. Lấy mẫu 50 nhân viên => lương của 50.
# Tính khoảng tin cậy 95% cho mức lương trung bình
n = 50
std_sal = 3.0
mean_sal = 18.0
salaries = np.random.normal(size=n, loc=mean_sal, scale=std_sal)
print(np.mean(salaries))
print(np.std(salaries))
```

```python
confidence = 0.95
alpha = 0.05
z = stats.norm.ppf(1-alpha/2)
print(z)
```

```python
margin = z * np.std(salaries) / np.sqrt(n)
print(margin)
```

```python
print(f'Mean salary is in {np.mean(salaries) - margin} - {np.mean(salaries) + margin} with confidence {confidence}%')
```

```python
# Khảo sát 20 công nhân, thời gian trung bình hoàn thành 1sp = 10p, độ lệch chuẩn là 2p
n = 20
mean_time = 10.0
std_time = 2.0
confidence = 0.95
alpha = 1 - confidence
t = stats.t.ppf(1-alpha/2, df=n-1)
print(t)
```

```python
margin = t * std_time / np.sqrt(n)
print(margin)
```

```python
print(f'Mean time for a product is {mean_time - margin} - {mean_time + margin} with confidence {confidence}%')
```

```python
# Khảo sát 200 khách hàng, 160 người hài lòng
# Với độ tin cậy 99% mức độ hài lòng là bao nhiêu?
n = 200
p = 160 / n
confidence = 0.99
alpha = 1 - confidence
z = stats.norm.ppf(1 - alpha/2)
margin = z * np.sqrt(p * (1-p) / n)
print(f'Tỉ lệ hài lòng nằm trong khoảng {p - margin} - {p + margin} với độ tin cậy {confidence}%')
```

```python
# Trường công bố điểm trung bình môn Toán = 7.5 => H0
# Làm thống kê, đưa H1: điểm trung bình môn Toán != 7.5
mean_declare = 7.5
# Lấy mẫu 25 học sinh
n_students = 25
# Mô phỏng dữ liệu mẫu
scores = np.random.normal(size=n_students, loc=7.0, scale=3)
mean_sample = np.mean(scores)
std_sample = np.std(scores)
print(f'Mean sample: {mean_sample:.2f}, std sample: {std_sample:.2f}')
```

```python
alpha = 0.01
t = (mean_sample - mean_declare) / (std_sample / np.sqrt(n_students))
print(f't-statistic: {t:.2f}')
```

```python
p_value = 2 * (1 - stats.t.cdf(abs(t), df=n_students-1))
print(f'p-value: {p_value:.4f}')
```

```python
if p_value < alpha:
    print('Bác bỏ giả thuyết H0')
else:
    print('Không bác bỏ giả thuyết H0')
```

```python
# Ví dụ: Nhà máy sản xuất phẩm, đo lường chất lượng bằng trọng lượng, y/c 500g, độ lệch chuẩn là 3g.
mean_weight = 500
std_weight = 3.0
n_samples = 100
# H0: nghi ngờ trọng lượng trung bình < 500

# Giả lập lấy mẫu
weights = np.random.normal(size=n_samples, loc=505, scale=std_weight)
mean_sample = np.mean(weights)
std_sample = np.std(weights)
print(f'Mean sample: {mean_sample:.2f}, std sample: {std_sample:.2f}')
```

```python
z = (mean_sample - mean_weight) / (std_sample / np.sqrt(n_samples))
print(f'z-statistic: {z:.2f}')
p_value = 1 - stats.norm.cdf(z)
print(f'p-value: {p_value:.4f}')
alpha = 0.05
if p_value < alpha:
    print('Bác bỏ giả thuyết H0')
else:
    print('Không bác bỏ giả thuyết H0')
```
