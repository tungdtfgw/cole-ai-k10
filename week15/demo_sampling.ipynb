```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({'Grade': np.random.randint(50, 100, size=1000)})
df.head()
```

```python
df.describe()
```

```python
# Lấy mẫu ngẫu nhiên
df_sample = df.sample(n=100, random_state=10)
df_sample.describe()
```

```python
df['Gender'] = np.random.choice(['Male', 'Female'], size=1000)
df.head()
```

```python
df_sample = df.sample(n=100, random_state=10)
df_sample.head()
```

```python
print(f'Số lượng Male tổng thể: {len(df[df["Gender"] == "Male"])}')
print(f'Số lượng Female tổng thể: {len(df[df["Gender"] == "Female"])}')
print(f'Số lượng Male mẫu: {len(df_sample[df_sample["Gender"] == "Male"])}')
print(f'Số lượng Female mẫu: {len(df_sample[df_sample["Gender"] == "Female"])}')
```

```python
df_male = df[df["Gender"] == "Male"]        # tầng nam
df_female = df[df["Gender"] == "Female"]    # tầng nữ
df_male_sample = df_male.sample(n=50)       # lấy ngẫu nhiên 50 nam
df_female_sample = df_female.sample(n=50)   # lấy ngẫu nhiên 50 nữ
df_sample_groups = pd.concat([df_male_sample, df_female_sample])
print(f'Số lượng Male mẫu: {len(df_sample_groups[df_sample_groups["Gender"] == "Male"])}')
print(f'Số lượng Female mẫu: {len(df_sample_groups[df_sample_groups["Gender"] == "Female"])}')
```

```python
df_sample_groups = df.groupby('Gender').apply(lambda x: x.sample(n=50, random_state=10))
print(f'Số lượng Male mẫu: {len(df_sample_groups[df_sample_groups["Gender"] == "Male"])}')
print(f'Số lượng Female mẫu: {len(df_sample_groups[df_sample_groups["Gender"] == "Female"])}')
```
