```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('USImport.csv', delimiter=';')
df.head()
```

```python
df.info()
```

```python
df['Quantity'] = df['Quantity'].str.replace(',', '.').astype(float).astype(int)
df['Weight'] = df['Weight'].str.replace(',', '.').astype(float)
df['Value (US$)'] = df['Value (US$)'].str.replace(',', '.').astype(float)
df.info()
```

```python
df.describe()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Quantity'], bins=50, kde=True)
plt.axvline(df['Quantity'].mean(), color='red', linestyle='dashed', linewidth=2, label='Mean')
plt.axvline(df['Quantity'].median(), color='green', linestyle='dashed', linewidth=2, label='Median')
plt.legend()
plt.title('Distribution of Quantity')
plt.xlabel('Quantity')
plt.ylabel('Frequency')
plt.show()
```

```python
print(f'Mean Quantity: {df["Quantity"].mean()}')
print(f'Median Quantity: {df["Quantity"].median()}')
print(f'Mode Quantity: {df['Quantity'].mode()[0]}')
print(f'Std Quantity: {df["Quantity"].std()}')
print(f'CV Quantity: {df["Quantity"].std() / df["Quantity"].mean()}')
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x=df['Quantity'])
plt.axvline(df['Quantity'].median(), color='red', linestyle='dashed', linewidth=2, label='Median')
plt.title('Boxplot of Quantity')
plt.xlabel('Quantity')
plt.show()
```

```python
# Làm tương tự với weight
plt.figure(figsize=(10, 6))
sns.histplot(df['Weight'], bins=20, kde=True)
plt.axvline(df['Weight'].mean(), color='red', linestyle='dashed', linewidth=2, label='Mean')
plt.axvline(df['Weight'].median(), color='green', linestyle='dashed', linewidth=2, label='Median')
plt.legend()
plt.title('Distribution of Weight')
plt.xlabel('Weight')
plt.ylabel('Frequency')
plt.show()
```

```python
# Làm tương tự với Value
# Làm tương tự với weight
plt.figure(figsize=(10, 6))
sns.histplot(df['Value (US$)'], bins=20, kde=True)
plt.axvline(df['Value (US$)'].mean(), color='red', linestyle='dashed', linewidth=2, label='Mean')
plt.axvline(df['Value (US$)'].median(), color='green', linestyle='dashed', linewidth=2, label='Median')
plt.legend()
plt.title('Distribution of Value (US$)')
plt.xlabel('Value (US$)')
plt.ylabel('Frequency')
plt.show()
```
