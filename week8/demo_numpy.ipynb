```python
import random as rd
a = np.array(rd.sample(range(100), 20))
print(a)
```

```python
evens = a[a%2 == 0]
print(evens)
```

```python
odds = a[a%2 == 1]
print(odds)
```

```python
a = np.array(rd.sample(range(-100, 100), 20))
print(a)
```

```python
# Lọc các số âm
negatives = a[a < 0]
print(negatives)
# Lọc các số dương
positives = a[a > 0]
print(positives)
```

```python
a = 100 * np.random.rand(20)
print(a)
print(max(a), min(a), sum(a), sum(a) / a.size, sep='\n')
```

```python
# Chuẩn hóa a về khoảng [0, 1]
m = max(a)
b = a / m
print(b)
```

```python
# standard scaler => trung bình = 0 và độ lệch chuẩn (std) = 1
s = np.std(a)
m = np.mean(a)
c = (a - m) / s
print(c)
print(np.mean(c), np.std(c))
```

```python
d = (a - a.min()) / (a.max() - a.min())
print(d)
```

```python
m1 = a.min()
m2 = a.max()
e = (a - m1) / (m2 - m1)
print(e)
```
