```python
products = ['macbook air', 'iphone', 'macbook pro', 'ipad', 'apple watch']
prices = [1500, 1000, 2000, 800, 400]
print(max(prices))
print(min(prices))
```

```python
max_index = prices.index(max(prices))
print(max_index)
most_expensive = products[max_index]
print(most_expensive)
```

```python
max_index = 0
max_price = prices[0]

for i in range(1, len(prices)):
    if prices[i] > max_price:
        # update max_price and max_index
        max_price = prices[i]
        max_index = i

print(max_index)
print(max_price)
```

```python
most_expensive = products[max_index]
print(most_expensive)
```

```python
count = 0
for i in range(len(prices)):
    if prices[i] >= 1000:
        count += 1
        print(f'{products[i]}: ${prices[i]}')

print(f'There are {count} products with price >= 1000')
```

```python
print(f'Sum of prices: ${sum(prices)}')
```

```python
sum_expensive = 0
for i in range(len(prices)):
    if prices[i] >= 1000:
        sum_expensive += prices[i]

print(f'Sum of prices >= 1000: ${sum_expensive}')
```

```python
# list comprehension
print(sum([price for price in prices if price >= 1000]))
```
