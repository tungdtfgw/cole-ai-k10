```python
en_vn = {
    'hello': 'xin chào',
    'goodbye': 'tạm biệt',
    'banana': 'quả chuối',
    'apple': 'quả táo'
}
print(en_vn)
```

```python
print(en_vn['hello'])
```

```python
print(en_vn['Hello'])
```

```python
en_word = input('Enter an English word: ')
if en_word in en_vn:
    print(f'Nghĩa của từ "{en_word}" là: {en_vn[en_word]}')
else:
    print(f'Từ "{en_word}" không có trong từ điển.')
```

```python
en_vn['hello'] = 'chào bạn' # modifying an existing key-value pair
print(en_vn)
```

```python
en_vn['thank you'] = 'cảm ơn' # adding a new key-value pair
print(en_vn)
```

```python
del en_vn['banana'] # deleting a key-value pair
print(en_vn)
```

```python
vn_word = en_vn.pop('goodbye') # removing a key-value pair and returning its value
print(en_vn)
print(vn_word)
```

```python
# for key, value in dictionary.items():
for en, vn in en_vn.items():
    print(f'Nghĩa của từ {en} là {vn}')
```

```python
for en in en_vn.keys():
    print(f'Nghia của từ {en} là {en_vn[en]}')
```

```python
print(en_vn.keys())
print(list(en_vn.keys()))
```

```python
print(list(en_vn.items()))
print(list(en_vn.values()))
```
