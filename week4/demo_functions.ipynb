```python
a = 10
b = 5

def swap(a, b):
    temp = a
    a = b
    b = temp
    print(f"Inside swap: a={a}, b={b}")

swap(a, b)
print(f"Outside swap: a={a}, b={b}") # a và b vẫn giữ giá trị ban đầu
```

```python
a = 10
b = 5

def swap(a, b):
    temp = a
    a = b
    b = temp

    print(f"Inside swap: a={a}, b={b}")
    return a, b

a, b = swap(a, b)
print(f"Outside swap: a={a}, b={b}") # a và b đã được hoán đổi giá trị
```

```python
x = 10
def double_x():
    global x    # khai báo biến x là biến toàn cục
    x = x * 2
    print(f'Inside double_x: x={x}')

double_x() # in ra 20
print(f'Outside double_x: x={x}') # in ra 20
```

```python
def factorial(n):
    if n == 0:
        return 1
    
    s = 1
    for i in range(1, n + 1):
        s *= i
    return s

print(f'0! = {factorial(0)}')
print(f'5! = {factorial(5)}')
print(f'-2! = {factorial(-2)}')
```

```python
def factorial(n):
    if n < 0:
        raise ValueError("Input must be a non-negative integer.")
        
    if n == 0:
        return 1
    
    s = 1
    for i in range(1, n + 1):
        s *= i
    return s

print(factorial(0))
print(factorial(5))
```

```python
print(factorial(-2))
```

```python
def draw_square(size):
    for _ in range(size):
        for _ in range(size):
            print('*', end=' ')
        print()

draw_square(3)
```

```python
draw_square(5)
```

```python
# Ex1: Viết hàm draw_square nhận thêm tham số ký tự để vẽ hình vuông, ví dụ draw_square(3, '#') sẽ vẽ hình vuông 3x3 với ký tự '#'
```

```python
def draw_square(size, c='*'):
    for _ in range(size):
        for _ in range(size):
            print(c, end=' ')
        print()
```

```python
draw_square(3)
```

```python
draw_square(3, '?')
```

Dưới 50 kWh: 1678 VND/kWh
Từ 51 đến 100 kWh: 1734 VND/kWh
Từ 101 đến 200 kWh: 2014 VND/kWh
Trên 200 kWh: 2536 VND/kWh
Viết hàm main sử dụng vòng lặp yêu cầu người dùng nhập số điện tiêu thụ, gọi hàm tính tiền rồi in ra giá phải trả, sau đó hỏi người dùng có muốn tiếp tục không, nếu có thì quay lại từ đầu hỏi người dùng nhập số điện.

```python
def main():
    calculating = True
    while calculating:
        consumption = int(input("Nhập số điện tiêu thụ (kWh): "))
        payment = cal_payment(consumption)
        print(f"Giá phải trả: {payment} VND")
        calculating = input("Có tiếp tục không? (y/n): ").lower() == 'y'

def cal_payment(consumption):
    if consumption <= 50:
        return consumption * 1678
    elif consumption <= 100:
        return 50 * 1678 + (consumption - 50) * 1734
    elif consumption <= 200:
        return 50 * 1678 + 50 * 1734 + (consumption - 100) * 2014
    else:
        return 50 * 1678 + 50 * 1734 + 100 * 2014 + (consumption - 200) * 2536

main()
```
