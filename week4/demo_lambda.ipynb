```python
import random as rd
numbers = rd.sample(range(-100, 100), 20)
print(numbers)
n = len(numbers)
print(n)
```

```python
print(max(numbers, key=lambda x: x if x > 0 else -x))
```

```python
abs = lambda x: x if x > 0 else -x
print(max(numbers, key=lambda x: abs(x) % 10))
```

```python
numbers = rd.sample(range(1, 100), 10)
print(numbers)
```

```python
double_numbers = list(map(lambda x: 2*x, numbers))
print(double_numbers)
```

```python
digits = list(map(lambda x: x%10, numbers))
print(digits)
```

```python
evens = list(filter(lambda x: x%2 == 0, numbers))
print(evens)
```

```python
odds = list(filter(lambda x: x%2 == 1, numbers))
print(odds)
```
