```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
```

```python
x = np.linspace(1, 10, 20)
y = 2 * x + np.random.normal(-5, 5, 20)

plt.figure(figsize=(10, 6))
plt.plot(x, y, 'o')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot')
plt.grid(True)
plt.show()
```

```python
r = stats.pearsonr(x, y)
print(f'Pearson correlation: {r[0]:.2f}')
```

```python
prices_changes = np.linspace(-20, 20, 20)
demands = 100 - 5 * prices_changes + np.random.normal(0, 20, 20)
```

```python
r = stats.pearsonr(prices_changes, demands)
print(f'Pearson correlation: {r[0]:.2f}')
```

```python
plt.figure(figsize=(10, 6))
plt.plot(prices_changes, demands, 'o')
plt.xlabel('Prices Changes')
plt.ylabel('Demands')
plt.title('Scatter Plot')
plt.grid(True)
plt.show()
```

```python
# Đọc file df.csv, clean dữ liệu, tính thử tương quan giữa các điểm SAT
```

```python
import pandas as pd
df_admission = pd.read_csv('df.csv')
df_admission.head()
```

```python
df_admission.dropna(inplace=True)
df_admission.isnull().sum()
```

```python
sat_scores = df_admission[['SATVR75', 'SATMT75']]
r = stats.pearsonr(sat_scores['SATVR75'], sat_scores['SATMT75'])
print(f'Pearson correlation: {r[0]:.2f}')
```

```python
plt.figure(figsize=(10, 6))
plt.plot(sat_scores['SATVR75'], sat_scores['SATMT75'], 'o')
plt.xlabel('SATVR75')
plt.ylabel('SATMT75')
plt.title('Scatter Plot 2 SAT scores')
plt.show()
```

```python
columns = ['Applications', 'Admissions', 'Admission Rate', 'Enrollment', 'SATVR75', 'SATMT75', 'ACTEN75', 'ACTMT75']
corr_matrix = df_admission[columns].corr()

plt.figure(figsize=(10, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix')
plt.show()
```

```python
X = np.random.exponential(scale=10, size=100000)
plt.figure(figsize=(10, 6))
sns.histplot(X, bins=50, kde=True)
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.title('Histogram of Exponential Distribution')
```

```python
from numpy.random.mtrand import sample
sample_size = 500
n = 500

means = []
for i in range(n):
    sample = np.random.choice(X, size=sample_size)
    mean = np.mean(sample)
    means.append(mean)
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(means, bins=50, kde=True)
plt.xlabel('Mean')
plt.ylabel('Frequency')
plt.title('Histogram of Sample Means')
```
