```python
vowels = 'aeiou'
for char in s:
    if char in vowels:
        print(char, end=' ')
print()
for char in s:
    if char not in vowels:
        print(char, end=' ')
```

```python
for i in range(m):
    for j in range(n):
        print(n, end=' ')
    print() # xuống dòng sau khi in xong 1 hàng
```

```python
for i in range(1, m+1):
    for j in range(n):
        print(i, end=' ')
    print() # xuống dòng sau khi in xong 1 hàng
```
