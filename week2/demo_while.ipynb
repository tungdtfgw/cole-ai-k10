```python
total = 0
while True:
    n = int(input('Enter n: '))
    if n < 0:
        break
    total += n

print('Total =', total)
```

```python
total = 0
negative = True

while negative:
    n = int(input('Enter n: '))
    if n >= 0:
        total += n
    else:
        negative = False

print('Total =', total)
```

```python
budget = 100
total = 0
buying = True

while buying:
    product = input('Enter product name: ')
    price = int(input('Enter price: '))
    if total + price <= budget:
        print('Bought', product)
        total += price
    else:
        print('Over budget. Stop buying.')
        buying = False

print('Total spent =', total)
```

```python
printing = True

while printing:
    n = int(input('Enter n: '))

    for i in range(1, n + 1):
        print(i, end=' ')

    print()

    answer = input('Continue printing? (y/n): ')
    if answer == 'n':
        printing = False
```
