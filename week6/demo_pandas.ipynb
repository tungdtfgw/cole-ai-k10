```python
df_training['Duration'].fillna(df_training['Duration'].min(), inplace=True)
df_training['Pulse'].fillna(df_training['Pulse'].max(), inplace=True)
df_training['Maxpulse'].fillna(df_training['Maxpulse'].mean(), inplace=True)
df_training['Calories'].fillna(df_training['Calories'].median(), inplace=True)
df_training.isnull().sum()
```

```python
def get_strength(duration):
  mean_duration = df_training['Duration'].mean()
  if duration > mean_duration:
    return 'Strong'
  else:
    return 'Weak'
```

```python
print(get_strength(80))
```

```python
print(get_strength(60))
```

```python
strength_col = df_training['Duration'].apply(get_strength)
strength_col
```

```python
df_training['Strength'] = strength_col
df_training
```

```python
df_training = pd.read_csv('datab7.csv')
df_training.dropna(inplace=True)
df_training.isnull().sum()
```
