```python
# number operators: +, -, *, /, //, %, **
a = 10
b = 3

print(f'{a} + {b} = {a + b}')
```

```python
print(f'{a} - {b} = {a - b}')
print(f'{a} * {b} = {a * b}')
print(f'{a} / {b} = {a / b}')  # float division
print(f'{a} // {b} = {a // b}')  # integer division
```

```python
print(f'{a} % {b} = {a % b}')  # remainder
```

```python
print(f'{a} ^ {b} = {a ** b}')  # exponentiation
```

```python
print(16 ** 0.5)  # square root
```

```python
# comparison operators: ==, !=, <, >, <=, >=
```

```python
a = 5
b = 10
c = 5
print(a == b)  # False
print(a != b)  # True
print(a < b)  # True
print(a > b)  # False
print(a <= c)  # True
print(a <= b)  # True
print(a >= c)  # True
print(a >= b)  # False
```

```python
a = True
b = False
print(a and b)  # False
print(a or b)  # True
print(not a)  # False
```

```python
age = int(input("Enter your age: "))
n_seats = int(input("Enter number of seats: "))

if age > 13 and n_seats > 0:
    print("You can watch the movie.")
    print("Enjoy the movie!")
else:
    print("You cannot watch the movie.")
```

```python
grade = int(input("Enter your grade: "))

if grade < 40:
    print('Failed')
elif grade < 60:
    print('Passed')
elif grade < 80:
    print('Merit')
else:
    print('Distinction')
```

```python
if grade >= 80:
    print('Distinction')
elif grade >= 60:
    print('Merit')
elif grade >= 40:
    print('Passed')
else:
    print('Failed')
```
