```python
# f-string: replace variables in string with their values via {}
```

```python
fruits = ['apple', 'banana', 'cherry', 'mango', 'orange']

for f in fruits:
    print(f'Fruit: {f}')
```

```python
student_scores = {
    'Alice': 85,
    'Bob': 92,
    'Charlie': 78,
    'David': 90,
}

for name, score in student_scores.items():
    print(f'{name}: {score}')

for name in student_scores:
    print(f'{name}: {student_scores[name]}')
```
