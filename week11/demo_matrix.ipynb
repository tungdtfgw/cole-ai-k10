```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(
    {
        'age': [25, 43, 26],
        'orders': [12, 3, 11],
        'payments': [100, 200, 220],
    },
    index=['a', 'b', 'c']
)
```

```python
A = np.array(
    [[1, 2, 3],
     [4, 5, 6]]
)
B = np.array(
    [[7, 8],
     [9, 10],
     [11, 12]]
)
print(A.shape, B.shape)
```

```python
C = A @ B
print(C)
print(C.shape)
```

```python
W0 = np.array(
    [[0.1, 0.2],
     [0.2, -0.3]]
)
b0 = np.array([0.5, -0.2])
x0 = np.array([3.0, 5.5])

Y0 = W0 @ x0 + b0
print(Y0)
print(W0.shape, x0.shape, b0.shape, Y0.shape)
```

```python
# M: là ma trận đặc trưng của các bài hát, cột 1 rock, cột 2 pop
M = np.array(
    [[0.9, 0.1],  # bài 1
     [0.1, 0.9],  # bài 2
     [0.3, 0.7],  # bài 3
     [0.2, 0.8],  # bài 4
     [0.4, 0.6]]  # bài 5
)
# U: ma trận sở thích của người dùng
U = np.array(
    [[0.7, 0.3, 0.4, 0.2, 0.5],  # người dùng A
     [0.2, 0.8, 0.8, 0.3, 0.3],  # người dùng B
     [0.4, 0.6, 0.1, 0.4, 0.8]]  # người dùng C
)

S = U @ M
print(S)
print(U.shape, M.shape, S.shape)
```

```python
square = np.array(
    [[0, 1, 1, 0, 0],
     [0, 0, 1, 1, 0]]
)
scale = np.array(
    [[0.5, 0.],
     [0., 2]]
)

rectange = scale @ square
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(square[0], square[1], 'o-')
ax.plot(rectange[0], rectange[1], 'o-')
plt.show()
```

```python
print(np.linalg.det(scale))
```

```python
t = np.deg2rad(45)
rotate = np.array(
    [[np.cos(t), -np.sin(t)],
    [np.sin(t), np.cos(t)]]
)
```

```python
rot_square = rotate @ square

fig, ax = plt.subplots(figsize=(5, 4))
ax.plot(square[0], square[1], 'o-')
ax.plot(rot_square[0], rot_square[1], 'o-')
plt.show()
```

```python
print(np.linalg.det(rotate))
```

```python
shearing = np.array(
    [[1, 2],
     [0, 1]]
)
sh_square = shearing @ square

fig, ax = plt.subplots(figsize=(15, 5))
ax.plot(square[0], square[1], 'o-')
ax.plot(sh_square[0], sh_square[1], 'o-')
plt.show()
```

```python
print(np.linalg.det(shearing))
```
