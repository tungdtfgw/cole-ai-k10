```python
class Student:
    # constructor of a class
    def __init__(self, name, age):
        self.name = name    # assign attribute value
        self.age = age
```

```python
# create a Student object
john = Student('John', 20)
# access attributes of john
print(john.name)
print(john.age)
```

```python
john.name = 'John Lennon'
john.age = 22

print(john.name, john.age)
```

```python
# class Student version 2
class Student:
    # constructor of a class
    def __init__(self, name, age):
        self.name = name    # assign attribute value
        self.age = age

    # a method
    def hi(self):
        print(f'Hi, I am a student. My name is {self.name}, I am {self.age} years old.')
```

```python
john = Student('John', 20)
paul = Student('Paul', 19)

john.hi()
paul.hi()
```

```python
john.name = ''  # public access cannot protect attribute from invalid value
john.age = -20
john.hi()
```

```python
# class Student version 3
class Student:
    def __init__(self, name, age):
        self.__name = name  # private attribute
        self.age = age      # public attribute
    
    def hi(self):
        print(f'Hi, I am a student. My name is {self.__name}, I am {self.age} years old.')
```

```python
john = Student('John', 20)
john.hi()
```

```python
print(john.age)
```

```python
print(john.__name)
```

```python
john.__name = 'Paul'
john.age = 50
john.hi()
```

```python
# class Student version 4
class Student:
    def __init__(self, name, age):
        self.__name = name      # private attribute
        self.__age = age        # private attribute
    
    # getters
    def get_name(self):
        return self.__name
    
    def get_age(self):
        return self.__age
    
    # setters
    def set_name(self, new_name):
        if new_name == '':
            raise ValueError('Name cannot be empty!')
        self.__name = new_name
    
    def set_age(self, new_age):
        if new_age < 0:
            raise ValueError('Age cannot be negative!')
        self.__age = new_age

    def hi(self):
        print(f'Hi, I am a student. My name is {self.__name}, I am {self.__age} years old.')
```

```python
john = Student('John', 20)
john.hi()
```

```python
print(john.get_name())
print(john.get_age())
```

```python
john.set_name('John Lennon')
john.set_age(25)
john.hi()
```

```python
john.set_name('')
```

```python
john.set_age(-20)
```

```python
# class Student version 5
class Student:
    def __init__(self, name, age):
        self.__name = name      # private attribute
        self.__age = age        # private attribute

    @property
    def name(self):             # property name
        return self.__name
    
    @property
    def age(self):              # property age
        return self.__age
    
    @name.setter
    def name(self, new_name):
        if new_name == '':
            raise ValueError('Name cannot be empty!')
        self.__name = new_name

    @age.setter
    def age(self, new_age):
        if new_age < 0:
            raise ValueError('Age cannot be negative!')
        self.__age = new_age
    
    def hi(self):
        print(f'Hi, I am a student. My name is {self.__name}, I am {self.__age} years old.')
```

```python
john = Student('John', 20)
john.hi()
```

```python
print(john.name, john.age)
john.name = 'John Lennon'
john.age = 25
john.hi()
```

```python
john.name = ''
```

```python
john.age = -10
```

```python
paul = Student('', -20)
paul.hi()
```
