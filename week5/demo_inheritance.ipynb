```python
class Person:
  def __init__(self, fname, lname):
    self.firstname = fname
    self.lastname = lname

  def hi(self):
    print(f'Hi, my name is {self.firstname} {self.lastname}')

p1 = Person('John', 'Doe')
p1.hi()
```

```python
class Student(Person):
  pass
```

```python
john = Student('John', 'Doe')
john.hi()
```

```python
print(john.firstname, john.lastname)
```

```python
class Student(Person):
  def __init__(self, fname, lname, age):
    super().__init__(fname, lname)
    self.age = age
```

```python
john = Student('John', 'Doe', 20)
john.hi()
```

```python
class Student(Person):
  def __init__(self, fname, lname, age):
    super().__init__(fname, lname)
    self.age = age

  # override method hi of super class
  def hi(self):
    super().hi()
    print(f'I am {self.age} years old!')
```

```python
john = Student('John', 'Doe', 20)
john.hi()
```
